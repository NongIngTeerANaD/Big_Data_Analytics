## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |

| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

# Part 1: Social Listening Warm-up
*งานเดี่ยว — ทุกคนต้องทำและส่งของตนเอง ไม่ต้องนำเสนอ (10 คะแนน)*

ใช้เพียง **1–3 ไฟล์** ก็พอ ไม่จำเป็นต้องโหลดทั้ง dataset

ฟิลด์ที่มี เช่น `text`, `created_at`, `author`, `uri`, `has_images`, `reply_to`, `predicted_language`

## ตอบ 4 คำถามหลัก

1. **คนโพสต์เมื่อไร:** จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
2. **ใช้ภาษาอะไร:** รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
3. **คนพูดถึงอะไร:** หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
4. **ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้:** สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด

> Part 1 เป็น warm-up ไม่ต้องทำ influencer analysis หรือ reply-network เว้นแต่สนใจทำเพิ่มเอง


```python
# ----------------- Your code here -----------------
# ใบ้: อ่าน JSON lines ทีละบรรทัดด้วย generator ไม่ต้องโหลดทั้งไฟล์
# def iter_jsonl(path):
#     with open(path, encoding='utf-8') as f:
#         for line in f:
#             yield json.loads(line)

# ===== 1.0 Load data (3 of 24 files, spread across the collection day) =====
!pip install -q fast-langdetect
from huggingface_hub import hf_hub_download

REPO_ID = "alpindale/two-million-bluesky-posts"
FILES = ["posts_20241127_075630.jsonl",   # morning UTC
         "posts_20241127_113858.jsonl",   # midday UTC
         "posts_20241127_152348.jsonl"]   # afternoon UTC

def iter_jsonl(path):
    # Stream one JSON object per line instead of loading the whole file
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                yield json.loads(line)

records = []
for fname in FILES:
    path = hf_hub_download(repo_id=REPO_ID, filename=fname,
                           repo_type="dataset", local_dir=DATA_DIR)
    for r in iter_jsonl(path):
        # keep only the fields needed; author is a pseudonymous DID, used only for counting
        records.append({
            "file": fname,
            "uri": r.get("uri"),
            "author": r.get("author"),
            "text": r.get("text") or "",
            "created_at": r.get("created_at"),
            "has_images": bool(r.get("has_images")),
            "is_reply": r.get("reply_to") is not None,
        })

bsky_raw = pd.DataFrame(records)
print("ไฟล์ที่ใช้:", FILES)
print("แถวดิบทั้งหมด:", f"{len(bsky_raw):,}")
print("ฟิลด์ที่มีจริงในไฟล์:", list(r.keys()), "(ไม่มี predicted_language → ต้อง detect ภาษาเอง)")

# ===== 1.1 Basic quality checks =====
bsky_raw["created_at"] = pd.to_datetime(bsky_raw["created_at"], utc=True,
                                        errors="coerce", format="ISO8601")
n_dup_uri   = bsky_raw["uri"].duplicated().sum()
n_bad_time  = bsky_raw["created_at"].isna().sum()
n_empty_txt = (bsky_raw["text"].str.strip() == "").sum()
n_dup_text  = bsky_raw.loc[bsky_raw["text"].str.strip() != "", "text"].duplicated().sum()

bsky = (bsky_raw.drop_duplicates("uri")
                .dropna(subset=["created_at"])
                .reset_index(drop=True))

print(f"uri ซ้ำ: {n_dup_uri:,} | created_at อ่านไม่ได้: {n_bad_time:,} | "
      f"ข้อความว่าง (เช่น รูปอย่างเดียว): {n_empty_txt:,} | ข้อความซ้ำกัน: {n_dup_text:,}")
print("จำนวนโพสต์หลัง clean:", f"{len(bsky):,}", "| ผู้โพสต์ไม่ซ้ำ:", f"{bsky['author'].nunique():,}")
print(f"สัดส่วน reply: {bsky['is_reply'].mean():.1%} | มีรูป: {bsky['has_images'].mean():.1%}")
bsky[["file", "created_at", "is_reply", "has_images", "text"]].head(3)
```

```python
# ----------------- Your code here -----------------
# ===== Q1: When do people post? =====
print("จำนวนโพสต์:", f"{len(bsky):,}")
print("created_at ต่ำสุด → สูงสุด (ทั้งหมด):", bsky["created_at"].min(), "→", bsky["created_at"].max())

# Posts stamped outside the collection day are back-dated / imported / clock errors
day0 = pd.Timestamp("2024-11-27", tz="UTC")
in_day = bsky["created_at"].between(day0, day0 + pd.Timedelta(days=1), inclusive="left")
print(f"created_at นอกวันที่ 27 พ.ย. 2024 (UTC): {(~in_day).sum():,} โพสต์ ({(~in_day).mean():.2%}) "
      "→ ไม่นำมาคิดเรื่องเวลา")
print(f"   - เก่ากว่าวันเก็บ (back-dated/import): {(bsky['created_at'] < day0).sum():,} | "
      f"ล้ำไปในอนาคต (นาฬิกาเครื่องผิด): {(bsky['created_at'] >= day0 + pd.Timedelta(days=1)).sum():,}")
old = bsky.loc[bsky["created_at"] < day0, "created_at"]
print(f"   - ในกลุ่มที่เก่ากว่า: อยู่ภายใน 1 วันก่อนหน้า {(old >= day0 - pd.Timedelta(days=1)).mean():.1%}, "
      f"เก่ากว่า 1 ปี {(old < day0 - pd.Timedelta(days=365)).mean():.1%}")
bt = bsky[in_day].copy()

# --- Window covered by each file (1st-99th percentile to ignore stragglers) ---
def q(p):
    return lambda s: s.quantile(p)
win = bt.groupby("file")["created_at"].agg(start=q(0.01), end=q(0.99), posts="size")
win["minutes"] = ((win["end"] - win["start"]).dt.total_seconds() / 60).round(1)
win["posts_per_min"] = (win["posts"] * 0.98 / win["minutes"]).round(0)
win["start_TH"] = win["start"].dt.tz_convert("Asia/Bangkok").dt.strftime("%H:%M")
win["end_TH"]   = win["end"].dt.tz_convert("Asia/Bangkok").dt.strftime("%H:%M")
win["start_US_ET"] = win["start"].dt.tz_convert("America/New_York").dt.strftime("%H:%M")
print("\nช่วงเวลาที่แต่ละไฟล์ครอบคลุม (ไฟล์ละ ~100k โพสต์ → ยิ่งใช้เวลาน้อย = คนโพสต์ถี่):")
display(win[["start_TH", "end_TH", "start_US_ET", "minutes", "posts", "posts_per_min"]])

# --- Posts by hour ---
bt["hour_utc"] = bt["created_at"].dt.hour
hourly = bt.groupby("hour_utc").size().rename("posts").to_frame()
hourly["hour_TH"] = (hourly.index + 7) % 24
hourly["share_%"] = (hourly["posts"] / hourly["posts"].sum() * 100).round(1)
print("จำนวนโพสต์รายชั่วโมง (แสดงเฉพาะชั่วโมงที่มี ≥ 1,000 โพสต์):")
display(hourly[hourly["posts"] >= 1000])
print(f"ชั่วโมงอื่น ๆ รวมกันมีเพียง {hourly.loc[hourly['posts'] < 1000, 'posts'].sum():,} โพสต์ (โพสต์ที่ถูกส่งเข้ามาช้า/เร็วกว่าช่วงเก็บ)")

peak_hour = hourly["posts"].idxmax()
busiest = win["posts_per_min"].idxmax()
print(f"ชั่วโมงที่มีโพสต์มากที่สุดในข้อมูล: {peak_hour:02d}:00 UTC (= {(peak_hour+7)%24:02d}:00 เวลาไทย), "
      f"{hourly.loc[peak_hour, 'posts']:,} โพสต์")
print(f"ช่วงที่โพสต์ถี่ที่สุด (posts/min): {busiest} ≈ {win.loc[busiest, 'posts_per_min']:,.0f} โพสต์/นาที "
      f"(เริ่ม {win.loc[busiest, 'start_TH']} น. เวลาไทย / {win.loc[busiest, 'start_US_ET']} น. US Eastern)")
```

### คำตอบ Q1 — คนโพสต์เมื่อไร

- ใช้ 3 จาก 24 ไฟล์ (ไฟล์ละ 100,000 โพสต์ เลือกให้กระจายเช้า/กลางวัน/บ่ายตามเวลา UTC) ได้ **299,999 โพสต์** หลังตัด uri ซ้ำ 1 รายการ จาก **145,897 บัญชี** (reply 49.1%, มีรูป 16.1%)
- ช่วงที่ข้อมูลครอบคลุมจริงคือ **27 พ.ย. 2024** เพียง 3 ช่วงสั้น ๆ (เวลาไทย) 14:56–15:26, 18:39–19:02 และ 22:23–22:39 รวม ~69 นาที ส่วน `created_at` ต่ำสุดย้อนไปถึงปี 2008 เพราะมีโพสต์ที่ถูก back-date/นำเข้าจากที่อื่น 20,624 โพสต์ (6.9%; 94% เก่ากว่า 1 ปี) จึงตัดออกจากการวิเคราะห์เวลา
- ชั่วโมงที่มีโพสต์มากที่สุดในข้อมูลคือ **22:00 น. เวลาไทย (15:00 UTC) = 92,269 โพสต์** แต่การนับรายชั่วโมงสะท้อนว่า "หยิบไฟล์ไหนมา" มากกว่าพฤติกรรมจริง ตัวชี้วัดที่ยุติธรรมกว่าคือ **อัตราโพสต์ต่อนาที**: 3,018 → 4,008 → **5,780 โพสต์/นาที** ช่วง 22:23 น. เวลาไทย (= 10:23 น. US Eastern) สูงเกือบ 2 เท่าของช่วงบ่ายไทย สอดคล้องกับฐานผู้ใช้หลักในอเมริกาที่เริ่มเข้าสู่ช่วงเช้า

```python
# ===== Q2: Which languages? =====
# The files have no predicted_language field, so detect it with fastText (fast-langdetect, lid.176 "lite" model)
from fast_langdetect import detect

URL_RE     = re.compile(r"https?://\S+|www\.\S+|\b[\w\-]+(?:\.[\w\-]+)+/\S*", re.I)
MENTION_RE = re.compile(r"@[\w.\-]+")

def clean_text(t):
    # Remove links and @handles so they do not bias language/phrase detection
    t = URL_RE.sub(" ", t)
    t = MENTION_RE.sub(" ", t)
    return re.sub(r"\s+", " ", t).strip()

def detect_lang(t, min_letters=3, min_score=0.5):
    t = clean_text(t)
    if len(re.sub(r"[\W\d_]", "", t)) < min_letters:   # empty / emoji-only / numbers-only
        return "und"
    try:
        res = detect(t, model="lite")
    except Exception:
        return "und"
    if isinstance(res, list):
        res = res[0] if res else {}
    return res.get("lang", "und") if res.get("score", 0) >= min_score else "und"

t0 = time.time()
bsky["lang"] = [detect_lang(t) for t in bsky["text"]]
print(f"detect ภาษา {len(bsky):,} โพสต์ ใช้เวลา {time.time()-t0:.0f} วินาที")

lang_tbl = bsky["lang"].value_counts().rename("posts").to_frame()
lang_tbl["share_%"] = (lang_tbl["posts"] / len(bsky) * 100).round(2)
lang_tbl["cum_%"] = lang_tbl["share_%"].cumsum().round(2)
known = bsky["lang"] != "und"
lang_tbl["share_of_identified_%"] = (lang_tbl["posts"] / known.sum() * 100).round(2)
if "und" in lang_tbl.index:
    lang_tbl.loc["und", "share_of_identified_%"] = np.nan
display(lang_tbl.head(12))
print(f"ระบุภาษาไม่ได้ (und: ว่าง/อีโมจิ/สั้นเกิน/ความมั่นใจต่ำ): {(~known).mean():.1%}")
print(f"จำนวนภาษาที่พบ: {bsky.loc[known, 'lang'].nunique()}")
top3 = lang_tbl["share_of_identified_%"].drop("und", errors="ignore").head(3)
print(f"3 ภาษาแรก ({', '.join(top3.index)}) ครอบคลุม {top3.sum():.1f}% ของโพสต์ที่ระบุภาษาได้")
print(f"ภาษาไทย (th): {(bsky['lang'] == 'th').sum():,} โพสต์ = {(bsky['lang'] == 'th').mean():.2%}")

# Sanity check by eye: one real example per top language
for lg in lang_tbl.index.drop("und", errors="ignore")[:6]:
    ex = bsky.loc[bsky["lang"] == lg, "text"].sample(1, random_state=RANDOM_SEED).iloc[0]
    print(f"[{lg}] {ex[:120]!r}")
```

### คำตอบ Q2 — ใช้ภาษาอะไร

- ไฟล์ไม่มี `predicted_language` จึง detect เองด้วย fastText (lid.176) หลังตัด URL/@mention ข้อความว่าง/อีโมจิ/สั้นเกิน/ความมั่นใจ < 0.5 จัดเป็น `und` (17.5%) และสุ่มตรวจด้วยตาแล้วว่าผลสมเหตุสมผล
- ภาษาหลัก (% ของโพสต์ทั้งหมด / % ของที่ระบุภาษาได้): **อังกฤษ 48.0% / 58.2%**, **ญี่ปุ่น 12.5% / 15.1%**, สเปน 4.2% / 5.1%, โปรตุเกส 3.0%, เยอรมัน 2.9%, ฝรั่งเศส 2.7%, เกาหลี 2.2% — พบทั้งหมด 114 ภาษา ภาษาไทยเพียง 0.41%
- 3 ภาษาแรก (en, ja, es) ครอบคลุม **78.4%** ของโพสต์ที่ระบุภาษาได้ → ถ้าทำคอนเทนต์บน Bluesky ควรรองรับ **อังกฤษเป็นหลัก + ญี่ปุ่นเป็นอันดับสอง** (ใหญ่กว่าสเปนเกือบ 3 เท่า) แล้วค่อยขยายไป es/pt/de/fr ตามตลาดเป้าหมาย ส่วนคอนเทนต์ภาษาไทยจะเจอผู้ชมน้อยมาก

```python
# ===== Q3: What are people talking about? =====
from nltk.corpus import stopwords
EN_STOP = set(stopwords.words("english"))
# Calendar words are capitalised but carry no topic, so treat them as stopwords for entity counting
CALENDAR = {"monday", "tuesday", "wednesday", "thursday", "friday", "saturday", "sunday",
            "january", "february", "march", "april", "may", "june", "july", "august",
            "september", "october", "november", "december", "jan", "feb", "mar", "apr",
            "jun", "jul", "aug", "sep", "sept", "oct", "nov", "dec", "im", "ive", "lol", "omg"}
ENT_STOP = EN_STOP | CALENDAR

# --- (a) Hashtags: count posts AND distinct authors (authors guard against one account spamming a tag) ---
tag_long = (bsky[["author", "text"]]
            .assign(tag=bsky["text"].str.findall(r"#(\w[\w\-]*)"))
            .explode("tag").dropna(subset=["tag"])
            .reset_index())
tag_long["tag"] = tag_long["tag"].str.lower()
tag_long = tag_long[~tag_long["tag"].str.fullmatch(r"\d+")].drop_duplicates(["index", "tag"])
has_tag = bsky.index.isin(tag_long["index"]).mean()
print(f"โพสต์ที่มี hashtag อย่างน้อย 1 อัน: {has_tag:.1%} → hashtag ใช้น้อย ต้องดูคำ/วลีในเนื้อความประกอบ")

tag_tbl = (tag_long.groupby("tag")
           .agg(posts=("index", "size"), authors=("author", "nunique"))
           .sort_values(["authors", "posts"], ascending=False))
tag_tbl["posts_per_author"] = (tag_tbl["posts"] / tag_tbl["authors"]).round(1)
print("\nTop 15 hashtag (เรียงตามจำนวนผู้โพสต์ไม่ซ้ำ):")
display(tag_tbl.head(15))

# --- (b) Copy-paste chains: many different authors posting (almost) the same text ---
en = bsky[bsky["lang"] == "en"].copy()
en["clean"] = en["text"].map(clean_text).str.lower()
en = en[en["clean"].str.len() >= 20]
en["prefix"] = en["clean"].str[:60]
chains = (en.groupby("prefix").agg(posts=("author", "size"), authors=("author", "nunique"))
            .query("authors >= 20").sort_values("authors", ascending=False))
print(f"\nข้อความแบบ copy-paste chain (ขึ้นต้นเหมือนกัน 60 ตัวอักษร, ≥20 ผู้โพสต์): {len(chains)} แบบ")
display(chains.head(5))

# --- (c) Named things people mention (capitalised words/phrases, not at sentence start) ---
ENT_RE = re.compile(r"\b[A-Z][a-zA-Z']+(?:\s+[A-Z][a-zA-Z']+){0,2}\b")
def entities(t):
    out = set()
    for m in ENT_RE.finditer(t):
        before = t[:m.start()].rstrip()
        if not before or before[-1] in ".!?:\"“\n":      # skip sentence-initial words
            continue
        words = [w for w in m.group().split() if w.lower() not in ENT_STOP and w.lower().replace("'", "") not in ENT_STOP]
        if words and len(words[0]) > 1:
            out.add(" ".join(words))
    return out

en_u = en.drop_duplicates("prefix")              # one copy per chain so chains do not dominate
ent_long = (en_u[["author", "text"]].assign(ent=en_u["text"].map(clean_text).map(entities))
            .explode("ent").dropna(subset=["ent"]))
ent_tbl = (ent_long.groupby("ent").agg(posts=("text", "size"), authors=("author", "nunique"))
           .sort_values("authors", ascending=False))
ent_tbl["share_of_en_%"] = (ent_tbl["posts"] / len(en_u) * 100).round(2)
print(f"\nชื่อเฉพาะ/วลีเด่นในโพสต์ภาษาอังกฤษ (n = {len(en_u):,} โพสต์ หลังตัด chain ซ้ำ):")
display(ent_tbl.head(20))

# --- (d) Real examples for the top items ---
def show_examples(pattern, k=2, label=None, case=False):
    m = bsky["text"].str.contains(pattern, case=case, regex=True)
    sub = bsky.loc[m, "text"].drop_duplicates()
    print(f"\n▶ {label or pattern}  (พบใน {m.sum():,} โพสต์)")
    for s in sub.sample(min(k, len(sub)), random_state=RANDOM_SEED):
        print("   -", s.replace("\n", " ")[:160])

for tag in tag_tbl.index[:3]:
    show_examples(r"#" + re.escape(tag) + r"\b", label="#" + tag)
for e in ent_tbl.index[:4]:
    show_examples(r"\b" + re.escape(e) + r"\b", label=e, case=True)
for pref in chains.index[:1]:
    show_examples(re.escape(pref[:40]), k=1, label="chain: " + pref[:40] + "...")
```

### คำตอบ Q3 — คนพูดถึงอะไร

- **Hashtag ใช้น้อย** มีเพียง **9.7%** ของโพสต์ จึงนับทั้งจำนวนโพสต์และจำนวนผู้โพสต์ไม่ซ้ำ (กันบัญชีเดียวสแปม) อันดับต้นเป็น **ชุมชนสร้างสรรค์และหนังสือ**: #art (702 ผู้โพสต์), #booksky (457), #books (333), #photography (301), #bookchallenge (294), #oc, #vtuber, #digitalart
